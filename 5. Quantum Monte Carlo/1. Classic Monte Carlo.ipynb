# 1. Markov Chains


> A sequence $X_1, X_2, ...$ of random elements of some set is a *Markov Chain* if the conditional distribution of $X_{n+1}$ given $X_1, X_2, ...$ depends on $X_n$ only. The set in which the $X_i$ take values is called the *state space* of the Markov chain.

‎ 


## 1.1. Basics

‎ 

Here we will try and explain all the terms in this initial Markov Chain definition. 

‎ 

- **Random Experiment** : This is any process or action whose result is uncertain.

    - Example 1: Flipping a coin.

    - Example 2: Rolling a six-sided die.

    - Example 3: Measuring the exact temperature in your room right now.

‎ 

- **Outcome** : An outcome is a single, specific result of a random experiment. It's one of the possible things that can happen.

    - Example 1: For the coin flip, one possible outcome is "Heads." Another is "Tails."

    - Example 2: For the die roll, one possible outcome is rolling a 4. Another is rolling a 1.

    - Example 3: For the temperature reading, one possible outcome is 21.3°C.

‎ 

- **State Space** : This is the set of all possible outcomes of a random experiment. It's the complete list of everything that could happen. We will denote it as $S$. 

    - Example 1: For the coin flip, the sample space is $S = \{ \text{Heads}, \text{Tails}\}$.

    - Example 2: For the die roll, the sample space is $S = \{ 1, 2, 3, 4, 5, 6 \}$.

    - Example 3: For the temperature reading, the sample space could be all real numbers between 15°C and 25°C, written as $S = [15.0, 25.0]$.

‎ 

- **Event** : An event is any collection of one or more outcomes from the sample space. It is a subset of the sample space that we are interested in.

    - **simple event** : Consists of just one outcome.

        - Example 1: For the die roll, the event "the die lands on 3" is a simple event. We can write this as $\{3\}$.

    - **compound event** : Consists of more than one outcome.

        - Example 2: For the die roll, the event "the outcome is an even number" is a compound event. It corresponds to $\{2, 4, 6\}$.

        - Example 3: For the die roll, the event "the outcome is greater than 4" is a compound event. It corresponds to $\{5, 6\}$.

‎ 

- **Random Element** (Random Variable) : A random variable is a rule that assigns a numerical value to each outcome in a sample space. It's a way to turn the results of a random experiment into numbers so we can do math with them. We will use capital letters like $X$ to represent random variables. 

    - Example 1: The random variable $X$ represents the outcome of the roll.

        - The sample space for $X$ is $S = \{ 1, 2, 3, 4, 5, 6 \}$.

        - An event can be described using the random variable. For instance, the event "rolling an even number" can be written as $X = {2, 4, 6}$ or $X = \text{Even}$.

‎ 

- **Probability** : measure of the likelihood that an event will occur. It is a number between 0 and 1, where 0 means the event is impossible and 1 means the event is certain. We denote the probability of an event $E$ as $P(E)$, and is calculated as:

    $$
    P(E) = \frac{\text{Number of outcomes in the Event}}{\text{Total number of outcomes in the State Space}}
    $$

    - Example 1 : The probability of rolling the number $5$ is: 
    
        $$
        P(X = 5) = \frac{\text{\# in \{5\}}}{\text{\# in \{1, 2, 3, 4, 5, 6\}}} = \frac{1}{6}
        $$

‎ 

- **Stochastic Process** : A Stochastic Process is a sequence of random elements, an ordered collection of random variables, usually indexed by time. It is displayed as $X_1, X_2, X_3, ...$, it could represent tracking a random phenomenon over time.

    - Example 4 - Daily Weather: Let $X_t$ be the maximum temperature on day $t$. The sequence $X_1, X_2, ..., X_{365}$ is a stochastic process describing the temperature over a year.

    - Example 5 - Repeated Coin Flips: Flipping a coin over and over again generates a sequence of random outcomes, $X_1, X_2, ..., X_n$, where each $X_i$ can be either Heads or Tails.

‎ 

- **Dependency** : This concept describes the relationship between different random variables in a sequence.

    - **Independence** : Two random variables are independent if the outcome of one has no influence whatsoever on the outcome of the other.

        - Example 6 - Rolling a die twice: Let $X_1$ and $X_2$ be the first two rolls. Knowing what you rolled on your first try $X_1 = x$ tells you nothing about what you'll get on the second roll, $X_1$ and $X_2$ are independent. This means: $P(X_2 = x_2|X_1 = x_1) = P(X_2 = x_2)$


    - **Dependence** : Two random variables are dependent if knowing the outcome of one changes the probabilities for the outcome of the other.

        - Example 7 - Deck without replacement: Let $X_1$ be the first card and $X_2$ the second. Since we are in a position without replacement, and the deck size is limited, then after we select the first event, then the total number of cards in the deck, the State Space will diminish, thus, the probability of getting the second card will be altered by the first measurement. 

‎ 

- **Conditional Probability** : This is the formal mathematical tool for handling dependence. The probability of an event happening given that we already know another event has happened. We define it as: 

    $$
    P(A|B) = \frac{P(A \cap B)}{P(B)}
    $$

    Meaning : the probability that $A$ occurs, given $B$ has occurred, is the ratio of the probability of both happening to the probability that $B$ happens. 

    - Example 8 - Rolling a die with condition: Event $A$ is "the outcome is even", and event $B$ is "the outcome is greater than 3". We have that the State Spaces are $S_A = A = \{2, 4, 6\}$ and $S_B = B = \{4, 5, 6\}$, thus, $S_{A \cap B} = A \cap B = \{4, 6\}$. Then: 

        $$
        P(A|B) = \frac{P(X \in \{4, 6\})}{P(X \in \{4, 5, 6\})} = \frac{2/6}{3/6} = \frac{2}{3}
        $$

    - Example 9 - Drawing cards without replacement: In a deck of 52 cards, say we have event $A$ be "the card is a heart" and event $B$ is "the card is red". Since we know that $A \cap B$ is "the card is a heart", since all hearts are red, then the probability $P(A|B)$ is given by:

        $$
        P(A|B) = \frac{P(\text{X = Heart})}{P(\text{X = Red})} = \frac{13/52}{26/52} = \frac{1}{2}
        $$

‎ 

- **Probability Distributions** : A distribution is a complete description of how probabilities are spread across a state space. It tells us how likely each outcome is.

    - **Discrete distributions**: Situations where the number of possible outcomes is finite or countably infinite. A probability mass function (PMF) assigns a probability to each possible outcome, and probabilities must add up to 1. The PMF is written as:

        $$
        P(X = x) \quad \text{for } x \in S \qquad \text{ and } \qquad \sum_{x \in S}P(X = x) = 1
        $$

        - Example 10 - Fair Die:  $P(X = k) = \frac{1}{6}, \quad k \in \{1, 2, 3, 4, 5, 6\}$

    - **Continuous distributions**: Situations where the outcome can take on any value within an interval or collection of intervals. The number of possibilities is uncountably infinite. Instead of PMFs, we use probability density functions (PDFs), probabilities are calculated as areas under the curve of the density and the total area under the PDF is 1. 

    $$
    P(a \leq x \leq b) = \int_{a}^{b} f(x)dx \qquad \text{ and } \qquad \int_{\infty}^{\infty}f(x) = 1
    $$

    - Example 11 - Uniform distribution for room temperature: Suppose the temperature is equally likely anywhere between $15 \degree C$ and $25 \degree C$, the probability density is given by the **uniform density**, given by: 

        $$
        f(x) = \frac{1}{(b - a)}, \qquad \text{ thus: } \qquad f(x) = \frac{1}{25 - 15} = 0.1, \quad 15 \leq x \leq 25
        $$

        And the probability the temperature is between $20 \degree C$ and $22 \degree C$ is: 
        $$
        P(20 \leq x \leq 22) = \int_{20}^{22} 0.1 \, dx = 0.2
        $$

‎ 

- **Conditional Distributions** : A Conditional Distribution is the complete probability distribution of a random variable, given that another random variable has taken on a specific, fixed value.

    - Discrete Case: Suppose $X$ and $Y$ are discrete random variables. The conditional distribution of X given Y = y is: 

        $$
        P(X = x | Y = y) = \frac{P(X = x, Y = y)}{P(Y = y)}
        $$ 

        Notice that $P(X = x | Y = y)$ still depends on $x$, thus, it still is a PMF or PDF, and if you plug an $x$ value, it becomes a conditional probability. 

    - Continuous Case : Suppose $X$ and $Y$ are continuous random variables. The conditional distribution of X given Y = y is defined using a conditional density function:

        $$
        f_{X|Y}(x|y) = \frac{f_{X, Y}(x, y)}{f_Y(y)}
        $$ 

        Where $f_{X, Y}(x, y)$ is the joint PDF of $X$ and $Y$ and $f_Y(y)$ is the marginal PDF of $Y$. 

‎ 

## 1.2 Markov Chain Explained

Now, with these definitions in hand, we can rewrite the Markov Chain definition in simpler terms: 

> A sequence $X_1, X_2, ...$ of random elements of some set is a *Markov Chain* if the conditional distribution of $X_{n+1}$ given $X_1, X_2, ...$ depends on $X_n$ only. The set in which the $X_i$ take values is called the *state space* of the Markov chain.

A Markov Chain is a stochastic process ($X_1, X_2, ...$) where the variables are dependent. This dependency is given by the conditional distribution such that the next state $X_{n+1}$ only depends on the present state $X_{n}$. This is written as: 

‎ 

$$
P(X_{n+1} = x_{n+1}|X_{n} = x_{n}, X_{n-1} = x_{n-1}, ..., X_1 = x_1) = P(X_{n+1} = x_{n+1}|X_{n} = x_{n})
$$

‎ 

Thus, this stochastic process has a very interesting propriety that the past steps give no additional information about the future of the sequence, it depends entirely on the present term. 

Markov Chains can have *stationary transition probabilities* if the conditional distribution of $X_{n+1}$ given $X_{n}$ does not depend on $n$, meaning that the evolution of the sequence doest change over time. This propriety is interesting for **MCMC** (Markov Chain - Monte Carlo)  algorithms, that aim to simulate single, fixed target distribution, which requires the underlying process to have stable, unchanging rules.
